```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

print("Libraries imported successfully.")
```

```python
CAPSTONE_PATH = "/content/drive/MyDrive/AIM_Capstone_AshokKumar_B240655ME/AIM_Capstone_AshokKumar_B240655ME"

print("Capstone folder exists:", os.path.exists(CAPSTONE_PATH))
```

```python
MATLAB_PATH = os.path.join(
    CAPSTONE_PATH,
    "MATLAB"
)

IITG_PATH = os.path.join(
    MATLAB_PATH,
    "IITG dataset"
)

DATASET_PATH = os.path.join(
    IITG_PATH,
    "cleaned_dataset"
)

DATA_PATH = os.path.join(
    DATASET_PATH,
    "data"
)

EXTRA_INFO_PATH = os.path.join(
    DATASET_PATH,
    "extra_infos"
)

METADATA_PATH = os.path.join(
    DATASET_PATH,
    "metadata.csv"
)

print("Paths defined successfully.")
```

```python
print("========== PATH VERIFICATION ==========")

print("MATLAB folder       :", os.path.isdir(MATLAB_PATH))
print("IITG dataset        :", os.path.isdir(IITG_PATH))
print("Dataset folder      :", os.path.isdir(DATASET_PATH))
print("Data folder         :", os.path.isdir(DATA_PATH))
print("Extra-info folder   :", os.path.isdir(EXTRA_INFO_PATH))
print("Metadata file       :", os.path.isfile(METADATA_PATH))
```

```python
print("========== DATASET STRUCTURE ==========\n")

for item in os.listdir(DATASET_PATH):

    item_path = os.path.join(DATASET_PATH, item)

    if os.path.isdir(item_path):
        print(f"[FOLDER] {item}")

    else:
        print(f"[FILE]   {item}")
```

```python
def count_files(folder_path):

    count = 0

    for root, dirs, files in os.walk(folder_path):
        count += len(files)

    return count


data_file_count = count_files(DATA_PATH)
extra_file_count = count_files(EXTRA_INFO_PATH)

print("========== FILE COUNTS ==========")

print("Files in data folder       :", data_file_count)
print("Files in extra_infos folder:", extra_file_count)
```

```python
data_extensions = []

for root, dirs, files in os.walk(DATA_PATH):

    for file in files:

        extension = os.path.splitext(file)[1].lower()

        if extension == "":
            extension = "[no extension]"

        data_extensions.append(extension)


extension_counts = Counter(data_extensions)

print("========== DATA FILE TYPES ==========\n")

for extension, count in extension_counts.items():

    print(f"{extension}: {count}")
```

```python
data_files = []

for root, dirs, files in os.walk(DATA_PATH):

    for file in files:

        data_files.append(
            os.path.join(root, file)
        )


print("Total data files:", len(data_files))

print("\nFirst 10 data files:\n")

for file in data_files[:10]:
    print(file)
```

```python
extra_files = []

for root, dirs, files in os.walk(EXTRA_INFO_PATH):

    for file in files:

        extra_files.append(
            os.path.join(root, file)
        )


print("Total extra-info files:", len(extra_files))

print("\nExtra-info files:\n")

for file in extra_files:
    print(file)
```

```python
missing_summary = pd.DataFrame({
    "Missing Count": metadata.isnull().sum(),
    "Missing Percentage": metadata.isnull().mean() * 100
})

missing_summary
```

```python
duplicate_count = metadata.duplicated().sum()

print("========== DUPLICATE CHECK ==========")

print("Number of exact duplicate rows:", duplicate_count)
```

```python
battery_counts = metadata["battery_id"].value_counts()

print("========== BATTERY COVERAGE ==========")

print("Number of unique batteries:",
      metadata["battery_id"].nunique())

print("\nRecords per battery:")
print(battery_counts)
```

```python
type_counts = metadata["type"].value_counts()

print("========== OPERATION TYPE COVERAGE ==========")

print(type_counts)
```

```python
plt.figure(figsize=(8, 5))

type_counts.plot(kind="bar")

plt.title("Records by Operation Type")
plt.xlabel("Operation Type")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

battery_counts.plot(kind="bar")

plt.title("Number of Records per Battery")
plt.xlabel("Battery ID")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
for column in ["Capacity", "Re", "Rct"]:

    if column in metadata.columns:

        print("\n================================")
        print(column)
        print("================================")

        print("Data type:",
              metadata[column].dtype)

        print("Missing values:",
              metadata[column].isna().sum())

        print("Empty-list values:",
              (metadata[column].astype(str).str.strip() == "[]").sum())

        print("Zero values:",
              (metadata[column].astype(str).str.strip() == "0").sum())
```

```python
existing_files = set()

for root, dirs, files in os.walk(DATA_PATH):

    for file in files:
        existing_files.add(file)


metadata_filenames = set(
    metadata["filename"]
    .dropna()
    .astype(str)
)


missing_files = metadata_filenames - existing_files

print("========== FILE REFERENCE CHECK ==========")

print("Unique filenames in metadata:",
      len(metadata_filenames))

print("Files referenced in metadata but not found:",
      len(missing_files))
```
