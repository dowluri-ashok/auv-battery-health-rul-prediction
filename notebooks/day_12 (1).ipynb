```python
import os
import numpy as np
import pandas as pd
import joblib

DATASET_PATH = "/content/drive/MyDrive/IITG dataset/cleaned_dataset"
V1_PATH = os.path.join(DATASET_PATH, "processed_dataset_v1")

data = np.load(
    os.path.join(V1_PATH, "sequences_v1.npz"),
    allow_pickle=True
)

X = data["X"]
y_soh = data["SoH"]
y_rul = data["RUL"]
battery_ids = data["battery_id"]
splits = data["split"]

print("Dataset Version 1 loaded successfully")
print("X shape:", X.shape)
print("SoH shape:", y_soh.shape)
print("RUL shape:", y_rul.shape)
```

```python
print("Number of unique batteries:")
print(len(np.unique(battery_ids)))
```

```python
print("Sequence Configuration")
print("-----------------------")
print("Sequence length:", X.shape[1])
print("Number of input features:", X.shape[2])
print("Number of SoH targets:", len(y_soh))
print("Number of RUL targets:", len(y_rul))
```

```python
import joblib
import os

scaler_path = os.path.join(V1_PATH, "feature_scaler_v1.pkl")

scaler = joblib.load(scaler_path)

print("Scaler loaded successfully")
print("Scaling method: Min-Max Scaling")
print("Feature range:", scaler.feature_range)
print("Number of features scaled:", scaler.n_features_in_)
```

```python
print("Input features used in Dataset V1:")
print("1. cycle_progress_scaled")
print("2. ambient_temperature_scaled")

print("\nTargets:")
print("1. SoH")
print("2. RUL")
```

```python
print("===== DATASET VERSION 1 FINAL VERIFICATION =====")

print("Total sequences:", len(X))
print("Sequence shape:", X.shape)
print("Unique batteries:", len(np.unique(battery_ids)))

print("\nSplit counts:")
print(pd.Series(splits).value_counts())

print("\nTarget check:")
print("SoH targets:", len(y_soh))
print("RUL targets:", len(y_rul))

print("\nDataset V1 status: VERIFIED")
```
