## 1. My Dataset – Sequence Findings

The processed sequence dataset contains battery degradation sequences
prepared for modelling.

The dataset contains:

- 2287 total sequences
- Sequence length of 10
- 2 input features
- SoH target
- RUL target

The sequences are divided into training, validation and test sets.

```python
# Dataset findings from the Week-2 sequence preparation

my_dataset = {
    "Total sequences": 2287,
    "Sequence length": 10,
    "Input features": 2,
    "Training sequences": 1516,
    "Validation sequences": 379,
    "Test sequences": 392
}

my_dataset_df = pd.DataFrame(
    list(my_dataset.items()),
    columns=["Parameter", "Value"]
)

display(my_dataset_df)
```

```python
# Sequence split counts from the prepared dataset

split_counts = pd.Series({
    "Train": 1516,
    "Validation": 379,
    "Test": 392
})

print("Sequence Split Distribution")
print("---------------------------")

display(
    split_counts.to_frame(name="Number of Sequences")
)
```

```python
plt.figure(figsize=(8, 5))

split_counts.plot(
    kind="bar"
)

plt.title("Train, Validation and Test Sequence Distribution")
plt.xlabel("Dataset Split")
plt.ylabel("Number of Sequences")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
